## 二、題目說明（Problem Description）

請設計一個 Python 程式，建立學生分數分析系統。

程式必須能夠接收多筆分數，根據指定的倍率（multiplier）及加分值（bonus）調整分數，接著計算調整後的分數平方、總分及平均分數，最後回傳分析結果。

### Part 1：完成函數

請閱讀以下函數，理解各部分的功能。

```python
def analyze_scores(*scores, **options):
    """
    分析學生分數。

    Parameters:
        *scores: 接收不定數量的分數
        **options: 接收選擇性設定，例如 multiplier 和 bonus

    Returns:
        一個字典，包含調整後的分數、平方、總分及平均分數。
    """
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)

    adjusted = [
        score * multiplier + bonus
        for score in scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }
```

### Part 3：思考與分析（Questions）

1. 在函數內部，`scores` 的資料型態是什麼？其值為何？
資料型態：tuple（元組）
其值：(60, 70, 80)
2. 在函數內部，`options` 的資料型態是什麼？其值為何？
資料型態：dict（字典）
其值：{'multiplier': 1.1, 'bonus': 5}
3. `score * multiplier` 與 `score ** 2` 有什麼不同？請說明兩個運算子的意義。
*（乘法運算子）：在這裡是用於計算乘法。score * multiplier 代表將學生的原始分數乘上指定的倍率（multiplier），用來放大或縮小分數的基準。

**（次方運算子）：在這裡是用於計算次方（冪次）。score ** 2 代表將調整後的分數取平方（即二次方），用來進行非線性的加權或轉換計算。
4. 為什麼函數使用 `return`，而不是只使用 `print()`？
print() 只是將結果印在螢幕上：電腦在執行完 print() 之後，該筆資料就直接消失了，其他程式碼無法再拿這筆資料來做後續的計算（例如加總、存入資料庫或進行條件判斷）。

return 可以將結果「回傳」給呼叫者：函數執行完畢後會把計算結果交給變數（例如你前面寫的 result），讓你可以對這筆資料進行後續的處理、分析或二次利用。
5. 如果呼叫 `analyze_scores(60, 70, 80)`，但沒有提供 `settings`，程式會如何執行？請說明原因。
沒有傳入任何關鍵字參數時，**options 會是一個空的字典 {}。此時呼叫 options.get("multiplier", 1)，因為字典裡找不到 "multiplier"，就會自動採用預設值 1；同理，bonus 也會採用預設值 0。因此分數經過調整後會維持原樣（乘以 1 再加 0），並順利完成後續的平方、總分與平均計算。

1. 原本的 `*scores` 與 `scores_list` 有什麼不同？
*scores（位置參數）：屬於位置參數。呼叫時不需要指定參數名稱，直接傳入多個數值（例如 analyze_scores(60, 70, 80)），並在函數內部被自動收集成一個 tuple。

scores_list（關鍵字參數）：屬於關鍵字參數。呼叫時必須明確寫出參數名稱，並傳入一個串列 (list) 物件（例如 scores_list=[60, 70, 80]）。
2. 為什麼原本的函數無法直接使用上述呼叫方式處理 `scores_list`？
因為原本的函數是設計用 *scores 來接收位置參數（也就是直接傳入的一系列數值
3. 請修改函數，使上述呼叫方式也能正確計算結果。
def analyze_scores(*scores, **options):
    """
    分析學生分數。
    支援透過 *scores 傳入位置參數，或透過 scores_list 關鍵字參數傳入串列。
    """
    multiplier = options.get("multiplier", 1)
    bonus = options.get("bonus", 0)
    
    # 從 options 中取出 scores_list，若沒有則預設為空串列
    scores_list = options.get("scores_list", [])
    
    # 將 *scores（tuple）轉換成串列，並與 scores_list 合併
    all_scores = list(scores) + list(scores_list)

    adjusted = [
        score * multiplier + bonus
        for score in all_scores
    ]

    squared = [
        score ** 2
        for score in adjusted
    ]

    total = sum(adjusted)
    average = total / len(adjusted) if adjusted else 0

    return {
        "adjusted_scores": adjusted,
        "squared_scores": squared,
        "total": total,
        "average": average
    }

## 三、語法整理（Learning Focus）

| 語法 | 用途 |
|---|---|
| `*scores`（函數定義） | 將不定數量的位置參數收集成 tuple |
| `**options`（函數定義） | 將不定數量的關鍵字參數收集成 dictionary |
| `score * multiplier` | 乘法 |
| `score ** 2` | 次方 |
| `f(*scores)` | 將可迭代物件解包為位置參數 |
| `f(**settings)` | 將字典解包為關鍵字參數 |
| `return` | 將函數的結果回傳給呼叫端 |
