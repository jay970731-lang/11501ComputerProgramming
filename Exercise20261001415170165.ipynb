問題：撰寫一Python程式，為使用者輸入年份（可能是西元或民國）找尋其相對應的十二生肖。十二生肖是以十二年為一週期，每一年分別以一種動物來表示猴（monkey）、雞（rooster）、狗(dog)、豬(pig)、鼠(rat)、牛(ox)、虎(tiger)、兔(rabbit)、龍(dragon)、蛇（snake）、馬(horse)，以及羊(sheep) 。

def get_zodiac(year_type, year_input):
    # 判斷年份類型並轉為西元年
    if year_type == "民國":
        ad_year = year_input + 1911
    elif year_type == "西元":
        ad_year = year_input
    else:
        return "無效的年份類型"

    # 生肖列表（依餘數排序：AD年 % 12）
    # 西元 0 年除以 12 餘 0，對應申猴；西元 4 年（餘 4）為子鼠
    zodiac_map = {
        0: "猴 (Monkey)",
        1: "雞 (Rooster)",
        2: "狗 (Dog)",
        3: "豬 (Pig)",
        4: "鼠 (Rat)",
        5: "牛 (Ox)",
        6: "虎 (Tiger)",
        7: "兔 (Rabbit)",
        8: "龍 (Dragon)",
        9: "蛇 (Snake)",
        10: "馬 (Horse)",
        11: "羊 (Sheep)",
    }

    # 計算餘數並返回結果
    remainder = ad_year % 12
    return zodiac_map[remainder]


def main():
    print("=== 十二生肖查詢系統 ===")
    print("1. 西元年")
    print("2. 民國年")

    choice = input("請選擇輸入類型 (1 或 2): ").strip()

    try:
        if choice == "1":
            year = int(input("請輸入西元年份（例如：2024）: "))
            zodiac = get_zodiac("西元", year)
            print(f"西元 {year} 年對應的生肖是：{zodiac}")
        elif choice == "2":
            year = int(input("請輸入民國年份（例如：113）: "))
            zodiac = get_zodiac("民國", year)
            print(f"民國 {year} 年（西元 {year + 1911} 年）對應的生肖是：{zodiac}")
        else:
            print("輸入錯誤，請選擇 1 或 2。")
    except ValueError:
        print("格式錯誤，請輸入有效的數字年份。")


if __name__ == "__main__":
    main()

將十進位小數轉換為二進位時，採用乘 2 取整數法（取出整數部分，將小數部分繼續乘以 2，直到小數部分為 0 或出現循環）：LaTeX 數學公式語法
$0.1 \times 2 = 0.2$ $\rightarrow$ 整數部分為 0，小數部分剩 $0.2$$0.2 \times 2 = 0.4$ $\rightarrow$ 整數部分為 0，小數部分剩 $0.4$$0.4 \times 2 = 0.8$ $\rightarrow$ 整數部分為 0，小數部分剩 $0.8$$0.8 \times 2 = 1.6$ $\rightarrow$ 整數部分為 1，小數部分剩 $0.6$$0.6 \times 2 = 1.2$ $\rightarrow$ 整數部分為 1，小數部分剩 $0.2$ （開始重複）$0.2 \times 2 = 0.4$ $\rightarrow$ 整數部分為 0 ...
觀察到的現象：十進位的有限小數 0.1，在二進位系統中會變成無限循環小數。

2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙
(1) IEEE 754 倍精度浮點數（Double Precision）格式Python 預設的 float 型態採用 IEEE 754 64 位元標準：1 位元：正負符號位元（Sign）11 位元：指數部分（Exponent）52 位元：尾數部分（Significand / Mantissa，隱藏首位的 1 後實際可儲存 53 位元精確度）二進位的 0.1 正規化後為：$1.100110011... \times 2^{-4}$為了塞入 52 位元的尾數空間，電腦必須將超過長度的位元捨去並進行捨入。
(2) 四捨六入五成雙（銀行家捨入法 / Round to Nearest, Ties to Even）IEEE 754 預設採用四捨六入五成雙的規則處理裁切點：若裁切位元小於一半，直接捨去。若裁切位元大於一半，向上進位。若裁切位元正好等於一半，則捨入到最近的偶數（Least Significant Bit 變成 0）。

5. 你的方法，對於一般兩個浮點數加法，在科學或金融上是好方法嗎？請思考。

金融領域：Decimal 是好方法，float 絕對不行評價：Decimal 是最佳選擇。
原因：零誤差要求：金融與帳務計算要求「絕對精確」，就算只差 $0.00000001$ 元也會導致報表對不起來甚至違法。法規規範：金融領域通常規定必須使用十進位進位規則（例如銀行家捨入法）。效能不敏感：金融交易重點在於精準與交易安全性，對微秒等級的 CPU 運算效能要求不如科學運算極端，因此 Decimal 雖然較慢，完全在可接受範圍內。

科學計算領域：Decimal 不是好方法，float + 誤差容忍度才是主流
評價：Decimal 不適用，繼續使用原生 float（配合容許誤差）才是正確做法。
原因：效能差異巨大：科學計算（如氣象模擬、物理模擬、機器學習）需要進行數百億次的矩陣運算。原生 float（IEEE 754）直接由 CPU/GPU 的硬體加速（FPU），而 Decimal 是透過軟體模擬，速度會慢 10 到 100 倍以上。
量測本身就有誤差：科學實驗中的數據本來就帶有誤差範圍（極限精度），追求「數學上的完全精確」在科學上沒有實質意義。
解決方式：科學界不使用 Decimal，而是透過提升浮點數精度（例如從 32 位元 float 提升至 64 位元 double），並在比對數據時使用 math.isclose() 或 NumPy 的 np.isclose() 來允許微小的微觀誤差（Epsilon）。
